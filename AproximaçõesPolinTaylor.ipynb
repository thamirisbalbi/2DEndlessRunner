```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def exp_taylor(n,x):
  fat = 1.0
  term = 1.0
  sum = term
  i = 1

  while i <=n:
    fat = fat * i
    term = term * x
    sum = sum + term/fat
    i = i+1
  return sum

```

```python
x = np.linspace(-2,2,50) #lim inf, lim sup, num de valores #testei valores positivos e negativos
y1 = exp_taylor(1,x) #grau do polin, polin

y2 = exp_taylor(2,x)
y3 = exp_taylor(3,x)
y4 = exp_taylor(4,x)
y5 = exp_taylor(5,x)

yOrig = np.exp(x) #plotando função que quero comparar


plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)
plt.plot(x,y4)
plt.plot(x,y5)
plt.plot(x,yOrig)
plt.show()
```
